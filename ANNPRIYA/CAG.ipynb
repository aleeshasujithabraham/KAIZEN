```python

from pydantic import BaseModel, Field
from typing import List


class ConceptNode(BaseModel):
    id: str
    name: str
    description: str
    difficulty: int = Field(ge=1, le=5)
    domain_tag: str


concept_nodes = []

for index, concept in enumerate(structured_result.concepts, start=1):
    node = ConceptNode(
        id=f"c{index:02d}",
        name=concept.name,
        description=concept.description,
        difficulty=concept.difficulty,
        domain_tag="Python Fundamentals"
    )

    concept_nodes.append(node)

print("Total Concept Nodes:", len(concept_nodes))
print("\nFirst Concept Node:")
print(concept_nodes[0].model_dump())
```

```python

from pydantic import BaseModel, Field


class PrerequisiteEdge(BaseModel):
    from_id: str
    to_id: str
    relation_type: str = "prerequisite"
    weight: float = Field(ge=1, le=10)


# Map concept names to their generated IDs
name_to_id = {
    concept.name.strip().casefold(): concept.id
    for concept in concept_nodes
}

edges = []

for concept in structured_result.concepts:
    target_id = name_to_id[concept.name.strip().casefold()]

    for prerequisite_name in concept.prerequisites:
        prerequisite_id = name_to_id.get(
            prerequisite_name.strip().casefold()
        )

        if prerequisite_id is None:
            print(
                "Warning: prerequisite not matched:",
                prerequisite_name,
                "for", concept.name
            )
            continue

        if prerequisite_id == target_id:
            print("Warning: self-dependency skipped:", concept.name)
            continue

        edges.append(
            PrerequisiteEdge(
                from_id=prerequisite_id,
                to_id=target_id,
                relation_type="prerequisite",
                weight=7
            )
        )

print("Total prerequisite edges:", len(edges))

for edge in edges[:5]:
    print(edge.model_dump())
```

```python

learning_graph = {
    "concepts": [
        node.model_dump()
        for node in concept_nodes
    ],
    "edges": [
        edge.model_dump()
        for edge in edges
    ]
}

print("Concepts:", len(learning_graph["concepts"]))
print("Edges:", len(learning_graph["edges"]))
print("\nFirst concept:")
print(learning_graph["concepts"][0])
```

```python

import networkx as nx

def build_learning_graph(syllabus):
    # 1. Analyze the supplied syllabus
    analyzed = structured_llm.invoke(
        syllabus_prompt.format_messages(syllabus=syllabus)
    )

    # 2. Detect the subject dynamically
    metadata = metadata_llm.invoke(
        metadata_prompt.format_messages(syllabus=syllabus)
    )
    domain = metadata.domain_tag

    # 3. Create concept nodes
    concept_nodes = []

    for index, concept in enumerate(analyzed.concepts, start=1):
        node = ConceptNode(
            id=f"c{index:02d}",
            name=concept.name,
            description=concept.description,
            difficulty=concept.difficulty,
            domain_tag=domain
        )
        concept_nodes.append(node)

    # 4. Map concept names to IDs
    name_to_id = {
        concept.name.strip().casefold(): concept.id
        for concept in concept_nodes
    }

    # 5. Create prerequisite edges
    edges = []

    for concept in analyzed.concepts:
        target_id = name_to_id[concept.name.strip().casefold()]

        for prerequisite in concept.prerequisites:
            source_id = name_to_id.get(prerequisite.strip().casefold())

            if source_id is None:
                print(
                    "Warning: unmatched prerequisite:",
                    prerequisite,
                    "for",
                    concept.name
                )
                continue

            if source_id == target_id:
                continue

            edges.append(
                PrerequisiteEdge(
                    from_id=source_id,
                    to_id=target_id,
                    relation_type="prerequisite",
                    weight=7
                )
            )

    # 6. Assemble the graph in the agreed format
    graph = {
        "concepts": [node.model_dump() for node in concept_nodes],
        "edges": [edge.model_dump() for edge in edges]
    }

    # 7. Validate graph references and cycles
    G = nx.DiGraph()
    concept_ids = {node["id"] for node in graph["concepts"]}

    for concept_id in concept_ids:
        G.add_node(concept_id)

    for edge in graph["edges"]:
        if (
            edge["from_id"] not in concept_ids
            or edge["to_id"] not in concept_ids
        ):
            raise ValueError(f"Invalid edge: {edge}")

        G.add_edge(edge["from_id"], edge["to_id"])

    if not nx.is_directed_acyclic_graph(G):
        raise ValueError("The learning graph contains a cycle.")

    return graph
```

```python

import networkx as nx

def validate_learning_graph(graph):
    concept_ids = {
        concept["id"] for concept in graph["concepts"]
    }

    errors = []
    seen_edges = set()
    G = nx.DiGraph()

    G.add_nodes_from(concept_ids)

    for edge in graph["edges"]:
        source = edge["from_id"]
        target = edge["to_id"]
        pair = (source, target)

        if source not in concept_ids or target not in concept_ids:
            errors.append(f"Edge references unknown concept: {pair}")
            continue

        if source == target:
            errors.append(f"Self-dependency found: {pair}")

        if pair in seen_edges:
            errors.append(f"Duplicate edge found: {pair}")

        seen_edges.add(pair)

        if edge["relation_type"] != "prerequisite":
            errors.append(f"Unexpected relation type: {pair}")

        if not 1 <= edge["weight"] <= 10:
            errors.append(f"Invalid weight: {pair}")

        G.add_edge(source, target)

    if not nx.is_directed_acyclic_graph(G):
        errors.append("The graph contains a circular dependency.")

    return errors
```

```python

import json

def score_prerequisite_edges(graph):
    if not graph["edges"]:
        return graph

    concept_names = {
        concept["id"]: concept["name"]
        for concept in graph["concepts"]
    }

    relationships = [
        {
            "from_id": edge["from_id"],
            "to_id": edge["to_id"],
            "prerequisite_name": concept_names[edge["from_id"]],
            "target_name": concept_names[edge["to_id"]]
        }
        for edge in graph["edges"]
    ]

    weighted_llm = llm.with_structured_output(WeightedRelationships)

    messages = edge_weight_prompt.format_messages(
        relationships=json.dumps(relationships, indent=2)
    )

    result = weighted_llm.invoke(messages)

    expected = {
        (edge["from_id"], edge["to_id"])
        for edge in graph["edges"]
    }

    returned = [
        (item.from_id, item.to_id)
        for item in result.relationships
    ]

    # Ensure every original relationship is returned exactly once.
    if len(returned) != len(expected) or set(returned) != expected:
        raise ValueError(
            "Gemini returned missing, duplicate, or unexpected relationships."
        )

    weight_lookup = {
        (item.from_id, item.to_id): item.weight
        for item in result.relationships
    }

    for edge in graph["edges"]:
        key = (edge["from_id"], edge["to_id"])
        edge["weight"] = weight_lookup[key]

    return graph
```

```python

def build_learning_graph(syllabus):
    # 1. Analyze the syllabus
    analyzed = structured_llm.invoke(
        syllabus_prompt.format_messages(syllabus=syllabus)
    )

    # 2. Detect the subject
    metadata = metadata_llm.invoke(
        metadata_prompt.format_messages(syllabus=syllabus)
    )
    domain = metadata.domain_tag

    # 3. Create concept nodes
    concept_nodes = []

    for index, concept in enumerate(analyzed.concepts, start=1):
        concept_nodes.append(
            ConceptNode(
                id=f"c{index:02d}",
                name=concept.name,
                description=concept.description,
                difficulty=concept.difficulty,
                domain_tag=domain
            )
        )

    # 4. Map names to IDs
    name_to_id = {
        concept.name.strip().casefold(): concept.id
        for concept in concept_nodes
    }

    # 5. Create prerequisite edges
    edges = []
    unmatched = []

    for concept in analyzed.concepts:
        target_id = name_to_id[concept.name.strip().casefold()]

        for prerequisite in concept.prerequisites:
            source_id = name_to_id.get(prerequisite.strip().casefold())

            if source_id is None:
                unmatched.append((concept.name, prerequisite))
                continue

            if source_id == target_id:
                unmatched.append((concept.name, prerequisite))
                continue

            edges.append(
                PrerequisiteEdge(
                    from_id=source_id,
                    to_id=target_id,
                    relation_type="prerequisite",
                    weight=7
                )
            )

    if unmatched:
        print("Warning: prerequisites could not be mapped:")
        for target, prerequisite in unmatched:
            print(f"  {prerequisite!r} -> {target!r}")

    graph = {
        "concepts": [node.model_dump() for node in concept_nodes],
        "edges": [edge.model_dump() for edge in edges]
    }

    # 6. Validate graph structure before scoring
    errors = validate_learning_graph(graph)

    if errors:
        raise ValueError("\n".join(errors))

    # 7. Score prerequisite relationships
    graph = score_prerequisite_edges(graph)

    # 8. Validate again
    errors = validate_learning_graph(graph)

    if errors:
        raise ValueError("\n".join(errors))

    return graph
```

```python

import json

def save_learning_graph(graph, filename="learning_graph.json"):
    # Validate before saving
    errors = validate_learning_graph(graph)

    if errors:
        raise ValueError(
            "Cannot save an invalid graph:\n" + "\n".join(errors)
        )

    with open(filename, "w", encoding="utf-8") as file:
        json.dump(graph, file, indent=2, ensure_ascii=False)

    print(f"Learning graph saved to: {filename}")
```

```python

from pydantic import BaseModel, Field
from typing import List, Literal

class EdgeReview(BaseModel):
    from_id: str
    to_id: str
    verdict: Literal["valid", "questionable", "invalid"]
    reason: str = Field(
        description="Explain why this prerequisite relationship is appropriate or problematic"
    )

class SemanticValidationReport(BaseModel):
    reviews: List[EdgeReview]
```

```python

from langchain_core.prompts import ChatPromptTemplate

semantic_validation_prompt = ChatPromptTemplate.from_messages([
    ("system", """
You are an expert curriculum designer reviewing a learning graph.

Review each provided prerequisite relationship.

A prerequisite edge points from the concept that should be learned
first to the concept that depends on it.

For every edge:
- Mark valid if the prerequisite is educationally justified.
- Mark questionable if it may help but is not strictly necessary.
- Mark invalid if the dependency is incorrect or unnecessarily restrictive.
- Explain your decision briefly.
- Do not add, remove, or modify relationships.
- Return exactly one review for every supplied edge.
- Preserve the supplied from_id and to_id exactly.
- Judge dependencies based on learning requirements, not merely
  the order in which topics appeared in the syllabus.
"""),
    ("human", """
Concepts:
{concepts}

Prerequisite edges to review:
{edges}
""")
])
```

```python

def review_graph_semantics(graph):
    concepts_json = json.dumps(graph["concepts"], indent=2)
    edges_json = json.dumps(graph["edges"], indent=2)

    review_llm = llm.with_structured_output(
        SemanticValidationReport
    )

    messages = semantic_validation_prompt.format_messages(
        concepts=concepts_json,
        edges=edges_json
    )

    report = review_llm.invoke(messages)

    expected = {
        (edge["from_id"], edge["to_id"])
        for edge in graph["edges"]
    }

    returned = [
        (review.from_id, review.to_id)
        for review in report.reviews
    ]

    if len(returned) != len(expected) or set(returned) != expected:
        raise ValueError(
            "Semantic review did not return exactly one review per edge."
        )

    return report
```

```python

valid_edges = []
questionable_edges = []
invalid_edges = []

for review in semantic_report.reviews:
    item = {
        "from_id": review.from_id,
        "to_id": review.to_id,
        "reason": review.reason
    }

    if review.verdict == "valid":
        valid_edges.append(item)
    elif review.verdict == "questionable":
        questionable_edges.append(item)
    elif review.verdict == "invalid":
        invalid_edges.append(item)

print("Valid relationships:", len(valid_edges))
print("Questionable relationships:", len(questionable_edges))
print("Invalid relationships:", len(invalid_edges))
```

```python

import networkx as nx

def validate_learning_graph(graph):
    concept_ids = {
        concept["id"] for concept in graph["concepts"]
    }
    errors = []
    seen_edges = set()
    G = nx.DiGraph()
    G.add_nodes_from(concept_ids)

    for edge in graph["edges"]:
        source = edge["from_id"]
        target = edge["to_id"]
        pair = (source, target)

        if source not in concept_ids or target not in concept_ids:
            errors.append(f"Unknown concept ID: {pair}")
            continue

        if source == target:
            errors.append(f"Self-dependency found: {pair}")

        if pair in seen_edges:
            errors.append(f"Duplicate edge found: {pair}")

        seen_edges.add(pair)

        if edge["relation_type"] != "prerequisite":
            errors.append(f"Unexpected relation type: {pair}")

        if not 1 <= edge["weight"] <= 10:
            errors.append(f"Invalid weight: {pair}")

        G.add_edge(source, target)

    if not nx.is_directed_acyclic_graph(G):
        errors.append("The graph contains a circular dependency.")

    return errors
```

```python

errors = validate_learning_graph(reviewed_java_graph)

if errors:
    print("Graph validation failed:")
    for error in errors:
        print("-", error)
else:
    print("Graph validation passed!")
    print("Total concepts:", len(reviewed_java_graph["concepts"]))
    print("Total prerequisite edges:", len(reviewed_java_graph["edges"]))
```
